### Function 1 – torch.tensor()

### Introduction

The `torch.tensor()` function is used to create a PyTorch tensor from existing data, such as Python lists, tuples, or numerical values.

A tensor is a data structure that can store numerical values in different dimensions.

For example:
- A single number is called a scalar (0-dimensional tensor).
- A list of numbers is called a vector (1-dimensional tensor).
- A table of numbers is called a matrix (2-dimensional tensor).

The `torch.tensor()` function is one of the fundamental functions in PyTorch because machine learning models use tensors to represent input data, weights, and other numerical information.

**Syntax:**

`torch.tensor(data, dtype=None, device=None)`

- `data`: The input data used to create the tensor.
- `dtype`: Specifies the data type, such as `torch.float32` or `torch.int64`.
- `device`: Specifies where the tensor is stored, such as CPU or GPU.

````python
# Example 1 - Working: Creating a 2D tensor

x = torch.tensor([[1, 2], [3, 4.]])

print(x)
````

x is the variable name.

torch.tensor() creates a tensor.

[[1, 2], [3, 4.]] is the input data.
The outer list contains two inner lists, so we get two rows.
Each row has two elements, so we get two columns.

*4. is a floating-point number. PyTorch converts all elements to a compatible floating-point data type.*

````python
# Example 2 - Working: Creating a tensor with float32 data type

y = torch.tensor([10, 20, 30, 40], dtype=torch.float32)

print(y)
print("Data type:", y.dtype)
print("Shape:", y.shape)
````

````python
# Example 3 - Breaking: Creating a tensor with unequal row lengths

try:
    z = torch.tensor([[1, 2], [3, 4, 5]])
    print(z)

except Exception as error:
    print("Error:", error)
````

### When to use torch.tensor()

The `torch.tensor()` function is used whenever numerical data needs to be converted into a PyTorch tensor. It is an essential function for preparing input data for machine learning and deep learning models.

In Adversarial Machine Learning, images, signals, and numerical features are represented as tensors before being processed by a neural network.

# Function 2 – torch.reshape()

### Introduction

The `torch.reshape()` function is used to change the shape or dimensions of an existing tensor without changing its elements.

It allows us to rearrange data into a required shape for further processing.

For example, a one-dimensional tensor can be converted into a two-dimensional tensor containing rows and columns.

**Syntax:**

`torch.reshape(input, shape)`

- `input`: The original tensor that needs to be reshaped.
- `shape`: The desired dimensions of the output tensor.

**Important:** The total number of elements must remain the same before and after reshaping.

````python
# Example 1 - Working: Reshape a 1D tensor into a 2D tensor

x = torch.tensor([1, 2, 3, 4, 5, 6])

y = torch.reshape(x, (2, 3))

print(y)
print("New shape:", y.shape)
````

````python
# Example 2 - Working: Reshape a 2D tensor into a 1D tensor

a = torch.tensor([[10, 20, 30], [40, 50, 60]])

b = torch.reshape(a, (6,))

print(b)
print("New shape:", b.shape)
````

````python
# Example 3 - Breaking: Trying to reshape into an incompatible shape

x = torch.tensor([1, 2, 3, 4, 5, 6])

try:
    y = torch.reshape(x, (2, 4))
    print(y)

except RuntimeError as error:
    print("Error:", error)
````

The basic rule of **reshaping**

***The total number of elements must remain exactly the same before and after reshaping.***

# Function 3 – torch.mean()

### Introduction

The `torch.mean()` function is used to calculate the arithmetic mean (average) of the elements in a tensor.

The mean is calculated by adding all the values and dividing the sum by the total number of elements.

**Example:**

Consider five numbers:

10, 20, 30, 40, 50

Mean = (10 + 20 + 30 + 40 + 50) / 5 = 30

In PyTorch, the `torch.mean()` function performs this calculation directly on tensor elements.

**Syntax:**

`torch.mean(input, dim=None)`

- `input`: The tensor whose mean needs to be calculated.
- `dim`: Specifies the dimension along which the mean is calculated. If omitted, the mean is calculated across all elements.

````python
# Understanding torch.mean()

x = torch.tensor([10.0, 20.0, 30.0, 40.0, 50.0])

result = torch.mean(x)

print("Mean:", result)
````

````python
# Example 1 - Working: Calculate the mean of all elements

x = torch.tensor([2.0, 4.0, 6.0, 8.0])

result = torch.mean(x)

print("Mean:", result)
````

The tensor contains four elements: 2, 4, 6, and 8.

The mean is calculated by adding all the elements and dividing the sum by the total number of elements.

Mean = (2 + 4 + 6 + 8) / 4 = 5.

The `torch.mean()` function calculates the average automatically and returns a tensor containing the result.

````python
# Example 2 - Working: Calculate mean along different dimensions

x = torch.tensor([[2.0, 4.0],
                  [6.0, 8.0]])

column_mean = torch.mean(x, dim=0)
row_mean = torch.mean(x, dim=1)

print("Column mean:", column_mean)
print("Row mean:", row_mean)
````

The example demonstrates how to calculate the mean along different dimensions of a 2D tensor.

The tensor contains two rows and two columns.

When `dim=0` is specified, PyTorch calculates the mean column-wise:
- First column: (2 + 6) / 2 = 4
- Second column: (4 + 8) / 2 = 6

When `dim=1` is specified, PyTorch calculates the mean row-wise:
- First row: (2 + 4) / 2 = 3
- Second row: (6 + 8) / 2 = 7

Therefore, the `dim` argument determines the direction along which the mean is calculated.

````python
# Example 3 - Breaking: Calculate mean using an integer tensor

x = torch.tensor([1, 2, 3, 4])

try:
    result = torch.mean(x)
    print(result)

except RuntimeError as error:
    print("Error:", error)
````

In this example, the tensor contains integer values. By default, PyTorch creates this tensor with an integer data type (`torch.int64`).

The `torch.mean()` function requires a floating-point or complex tensor for this operation. Therefore, it raises a RuntimeError.

This problem can be resolved by converting the integer tensor into a floating-point tensor using the `.float()` method.

For example:

`torch.mean(x.float())`

This converts the tensor values into floating-point numbers before calculating the mean.

#####Understanding the dim Parameter in torch.mean()

The `dim` parameter specifies the dimension along which the mean (average) is calculated.

**1. For a 1D Tensor**

A 1D tensor contains a single dimension.

Example:

`x = torch.tensor([10.0, 20.0, 30.0, 40.0])`

- `torch.mean(x)` calculates the average of all elements.
- `torch.mean(x, dim=0)` also calculates the average of all elements.

Both produce the same result: `tensor(25.)`.

**2. For a 2D Tensor**

A 2D tensor contains rows and columns.

Example:

`x = torch.tensor([[2.0, 4.0], [6.0, 8.0]])`

- `torch.mean(x, dim=0)` calculates the mean column-wise.
  - First column: (2 + 6) / 2 = 4
  - Second column: (4 + 8) / 2 = 6
  - Output: `tensor([4., 6.])`

- `torch.mean(x, dim=1)` calculates the mean row-wise.
  - First row: (2 + 4) / 2 = 3
  - Second row: (6 + 8) / 2 = 7
  - Output: `tensor([3., 7.])`

- `torch.mean(x, dim=2)` produces an error because a 2D tensor has only dimensions 0 and 1.

**3. For a 3D Tensor**

A 3D tensor has three dimensions: 0, 1, and 2.

For example, a tensor with shape `(2, 2, 2)` supports:

- `torch.mean(x, dim=0)`
- `torch.mean(x, dim=1)`
- `torch.mean(x, dim=2)`

Each operation calculates the mean along the specified dimension.

**Important Points to Remember**

| Tensor | Available dimensions |
|---|---|
| 1D | dim=0 |
| 2D | dim=0, dim=1 |
| 3D | dim=0, dim=1, dim=2 |
| 4D | dim=0, dim=1, dim=2, dim=3 |

**Note:** PyTorch starts counting dimensions from zero. The dimension specified in `dim` is reduced (removed) when calculating the mean.

# Function 4 – torch.sign()

### Introduction

The `torch.sign()` function returns the sign of each numerical element in a tensor.

It identifies whether a value is positive, negative, or zero.

The function follows these rules:

- Positive value → 1
- Negative value → -1
- Zero → 0

**Syntax:**

`torch.sign(input)`

- `input`: The tensor whose element signs need to be calculated.

The `torch.sign()` function is useful in mathematical operations and machine learning algorithms. In Adversarial Machine Learning, it plays an important role in the Fast Gradient Sign Method (FGSM), where the sign of the gradient is used to determine the direction of input perturbation.

````python
# Understanding torch.sign()

x = torch.tensor([-5.0, -1.0, 0.0, 2.0, 7.0])

result = torch.sign(x)

print("Original values:", x)
print("Sign values:", result)
````

The example demonstrates how to find the sign of each element in a tensor.

The `torch.sign()` function checks every element and returns:
- -1 for negative values.
- 0 for zero.
- 1 for positive values.

The output contains the signs of the original tensor elements instead of their actual magnitudes.

````python
# Example 2 - Working: Find the sign of gradient values

gradient = torch.tensor([-0.8, 0.3, -0.1, 0.0, 1.2])

result = torch.sign(gradient)

print("Gradient:", gradient)
print("Sign of gradient:", result)
````

This example demonstrates how `torch.sign()` can be applied to gradient values.

A gradient indicates how the loss changes with respect to a particular variable.

The `torch.sign()` function extracts the direction of each gradient value:
- Negative gradient → -1
- Positive gradient → 1
- Zero gradient → 0

In the Fast Gradient Sign Method (FGSM), the sign of the gradient of the loss with respect to the input is used to determine the direction of the adversarial perturbation.

Therefore, `torch.sign()` is an important operation in gradient-based adversarial attacks.

````python
# Example 3 - Breaking: Applying sign to string values

try:
    x = torch.tensor(["hello", "world"])
    result = torch.sign(x)
    print(result)

except (TypeError, ValueError, RuntimeError) as error:
    print("Error:", error)
````

The `torch.sign()` function performs a mathematical operation and requires numerical tensor values.

In this example, strings such as "hello" and "world" are provided instead of numerical values.

PyTorch cannot create a regular numerical tensor from these strings, so an error occurs.

This demonstrates that `torch.sign()` is intended for numerical tensor operations and cannot be applied to ordinary text strings.

# Function 5 – torch.clamp()

### Introduction

The `torch.clamp()` function is used to restrict the values of a tensor within a specified minimum and maximum range.

If a value is smaller than the specified minimum, it is replaced by the minimum value.

If a value is greater than the specified maximum, it is replaced by the maximum value.

Values that are already within the specified range remain unchanged.

### Syntax

```python
torch.clamp(input, min=None, max=None)
```

**Parameters:**

- `input`: The input tensor.
- `min`: The minimum permitted value.
- `max`: The maximum permitted value.

### Why is torch.clamp() useful?

In image processing, pixel values are commonly represented within the range 0 to 1 or 0 to 255.

When modifications are applied to image pixels, some values may become smaller than the minimum or greater than the maximum permitted value.

`torch.clamp()` helps restrict these values to the valid range.

This is particularly useful in Adversarial Machine Learning, where small perturbations are added to images to test model robustness.


````python
import torch

x = torch.tensor([-5.0, 0.5, 2.0, 5.0, 10.0])

result = torch.clamp(x, min=0.0, max=5.0)

print("Original tensor:", x)
print("Clamped tensor:", result)
````

````python
x = torch.tensor([-10.0, -2.0, 0.0, 3.0, 8.0])

minimum_result = torch.clamp(x, min=0.0)
maximum_result = torch.clamp(x, max=5.0)

print("Original tensor:", x)
print("Minimum clamp:", minimum_result)
print("Maximum clamp:", maximum_result)
````

````python
x = torch.tensor([1.0, 2.0, 3.0, 4.0])

try:
    result = torch.clamp(x, min=5.0, max=2.0)
    print(result)

except RuntimeError as error:
    print("Error:", error)

except Exception as error:
    print("Error:", error)
````

## Important Note: Understanding `torch.clamp()`

### 1. Normal usage

```python
x = torch.tensor([1.0, 3.0, 8.0])
result = torch.clamp(x, min=2.0, max=5.0)
print(result)
```

Output:
```text
tensor([2., 3., 5.])
```

**Rules:**
- Values below the minimum become the minimum.
- Values above the maximum become the maximum.
- Values within the range remain unchanged.

### 2. What happens when minimum is greater than maximum?

```python
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
result = torch.clamp(x, min=5.0, max=2.0)
print(result)
```

Output:
```text
tensor([2., 2., 2., 2.])
```

**Important:** When `min > max`, PyTorch sets every output element to the maximum value. It does not raise an error.

### 3. Remember

| Condition | Result |
|---|---|
| `min < max` | Normal clamping |
| `min > max` | All elements become `max` |
| Only `min` specified | Values below minimum are replaced |
| Only `max` specified | Values above maximum are replaced |

### 4. Application in Adversarial Machine Learning

After adding adversarial perturbations to an image, pixel values may go outside the valid range.

For images normalized between 0 and 1:

```python
image = torch.clamp(image, min=0.0, max=1.0)
```

This ensures that all pixel values remain between 0 and 1.

**Key takeaway:** `torch.clamp()` restricts tensor values to a specified range and helps maintain valid numerical and image pixel values.


#Understanding Autograd in PyTorch

## 1. Introduction

Autograd stands for **Automatic Differentiation**.

It is a feature of PyTorch that automatically calculates gradients (derivatives) of mathematical operations performed on tensors.

In machine learning, gradients help models understand how their parameters should change to reduce prediction errors.

Instead of calculating derivatives manually, PyTorch uses Autograd to calculate them automatically.

## 2. Why is Autograd Needed?

Consider the mathematical equation:

$$
y=x^2
$$

If the input value is:

$$
x=3
$$

Then:

$$
y=3\times3=9
$$

To understand how the output changes with respect to the input, we calculate the derivative.

$$
\frac{dy}{dx}=2x
$$

Substituting x = 3:

$$
\frac{dy}{dx}=2\times3=6
$$

Therefore, the gradient is 6.

**Interpretation:** At x = 3, the rate of change of y with respect to x is 6.

Autograd calculates this gradient automatically using PyTorch.

## 3. Important Autograd Concepts

### A. requires_grad=True

This tells PyTorch to track mathematical operations involving a tensor so that gradients can be calculated.

Example:

```python
x = torch.tensor(3.0, requires_grad=True)
```

### B. backward()

The `backward()` function calculates gradients by propagating derivatives backward through the computational graph.

Example:

```python
y.backward()
```

### C. .grad

The `.grad` attribute stores the calculated gradient of a tensor.

Example:

```python
print(x.grad)
```

### D. grad_fn

The `grad_fn` attribute indicates the operation that created a tensor and connects it to the computational graph.

Example:

```python
y = x * x
print(y.grad_fn)
```

## 4. What is a Computational Graph?

A computational graph is a representation of mathematical operations performed to calculate an output.

Consider:

$$
x=3
$$

$$
y=x\times x
$$

The computational graph tracks:

- Input: x
- Operation: Multiplication
- Output: y

PyTorch uses this graph to calculate gradients when `backward()` is called.

## 5. How Autograd Works

1. Create a tensor with `requires_grad=True`.
2. Perform mathematical operations using the tensor.
3. PyTorch tracks these operations in a computational graph.
4. Call `backward()` to calculate gradients.
5. Access the calculated gradient using `.grad`.

## 6. Example 1 – Basic Gradient Calculation

Consider:

$$
y=x^2
$$

The derivative is:

$$
\frac{dy}{dx}=2x
$$

For x = 3, the gradient is 6.

### Python Code

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x * x

y.backward()

print("Gradient:", x.grad)
```

**Output:**

```text
Gradient: tensor(6.)
```

**Explanation:**

- `torch.tensor()` creates a tensor.
- `requires_grad=True` enables gradient tracking.
- `y = x * x` performs multiplication.
- `y.backward()` calculates the gradient.
- `x.grad` displays the calculated gradient.

## 7. Example 2 – Gradient Calculation with Multiple Inputs

Autograd can calculate gradients for multiple tensors.

Consider:

$$
z=x^2+y^2
$$

The partial derivative with respect to x is:

$$
\frac{\partial z}{\partial x}=2x
$$

The partial derivative with respect to y is:

$$
\frac{\partial z}{\partial y}=2y
$$

For x = 2 and y = 3:

$$
\frac{\partial z}{\partial x}=2(2)=4
$$

$$
\frac{\partial z}{\partial y}=2(3)=6
$$

### Python Code

```python
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)

z = x**2 + y**2

z.backward()

print("Gradient of x:", x.grad)
print("Gradient of y:", y.grad)
```

**Output:**

```text
Gradient of x: tensor(4.)
Gradient of y: tensor(6.)
```

## 8. Example 3 – Gradient Accumulation and Resetting

In PyTorch, gradients accumulate by default when `backward()` is called multiple times.

To clear previously calculated gradients, we use the `zero_()` method.

### Python Code

```python
x = torch.tensor(3.0, requires_grad=True)

y = x * x
y.backward()

print("First gradient:", x.grad)

x.grad.zero_()

print("Gradient after resetting:", x.grad)
```

**Output:**

```text
First gradient: tensor(6.)
Gradient after resetting: tensor(0.)
```

**Explanation:**

- First, the gradient is calculated as 6.
- `x.grad.zero_()` resets the stored gradient to zero.
- This is important during neural network training because gradients generally need to be cleared before the next training step.

## 9. Role of Gradients in Machine Learning

During model training:

1. The model receives input data.
2. It produces a prediction.
3. A loss function calculates the prediction error.
4. `backward()` calculates gradients.
5. An optimizer uses the gradients to update model parameters.
6. The process repeats to improve the model.

A basic gradient descent update is:

$$
w_{\text{new}}=w_{\text{old}}-\eta\frac{dL}{dw}
$$

Where:

- w = Model weight.
- L = Loss function.
- η = Learning rate.
- dL/dw = Gradient of loss with respect to the weight.

## 10. Application in Adversarial Machine Learning

Gradients are also important in Adversarial Machine Learning.

An adversarial attack introduces small, carefully chosen changes to input data to mislead a machine learning model.

One example is the Fast Gradient Sign Method (FGSM).

It uses the sign of the gradient of the loss with respect to the input.

$$
x_{\text{adv}}=x+\epsilon\cdot\operatorname{sign}(\nabla_x L)
$$

Where:

- x = Original input.
- x_adv = Adversarial input.
- ε = Perturbation magnitude.
- L = Loss function.
- ∇x L = Gradient of loss with respect to input.
- sign = Extracts the sign of each gradient value.

## 11. Important PyTorch Autograd Functions

| PyTorch code | Purpose |
|---|---|
| `requires_grad=True` | Enables gradient tracking |
| `tensor.grad` | Retrieves the calculated gradient |
| `tensor.grad_fn` | Shows the operation that created a tensor |
| `backward()` | Calculates gradients |
| `tensor.grad.zero_()` | Resets a tensor's gradient |
| `optimizer.zero_grad()` | Clears optimizer-managed gradients |
| `torch.no_grad()` | Disables gradient tracking within its context |
| `torch.autograd.grad()` | Calculates and returns gradients directly |

Autograd is an important feature of PyTorch that automatically calculates gradients using automatic differentiation.

The `requires_grad=True` parameter enables gradient tracking, `backward()` calculates gradients, and `.grad` stores the calculated values.

Autograd plays an important role in deep learning optimization and adversarial machine learning, where gradients are used to understand and manipulate model behavior.
